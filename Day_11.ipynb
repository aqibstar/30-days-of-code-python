```python
%%writefile student_module.py

FILE_NAME = "students.txt"

def load_students():
    """Read all students from file into a dictionary."""
    students = {}
    try:
        f = open(FILE_NAME, "r")
        lines = f.readlines()
        f.close()

        for line in lines:
            line = line.strip()
            if line:
                roll, name, marks = line.split(",")
                students[roll] = {"name": name, "marks": int(marks)}

    except FileNotFoundError:
        pass

    return students


def save_students(students):
    """Save dictionary of students back to file."""
    f = open(FILE_NAME, "w")
    for roll, info in students.items():
        f.write(f"{roll},{info['name']},{info['marks']}\n")
    f.close()


def calculate_grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 60:
        return "C"
    elif marks >= 40:
        return "D"
    else:
        return "F"


def add_student():
    students = load_students()
    roll = input("Enter Roll Number: ")

    if roll in students:
        print("Student with this roll number already exists!")
        return

    name = input("Enter Name: ")
    marks = int(input("Enter Marks: "))

    students[roll] = {"name": name, "marks": marks}
    save_students(students)
    print("Student added successfully!")


def view_students():
    students = load_students()
    if not students:
        print("No students found.")
        return

    print("\nRoll No | Name | Marks | Grade")
    print("-" * 35)
    for roll, info in students.items():
        grade = calculate_grade(info["marks"])
        print(f"{roll} | {info['name']} | {info['marks']} | {grade}")


def search_student():
    students = load_students()
    roll = input("Enter Roll Number to search: ")

    if roll in students:
        info = students[roll]
        grade = calculate_grade(info["marks"])
        print(f"Found: {info['name']} | Marks: {info['marks']} | Grade: {grade}")
    else:
        print("Student not found!")


def update_student():
    students = load_students()
    roll = input("Enter Roll Number to update: ")

    if roll in students:
        name = input("Enter new name: ")
        marks = int(input("Enter new marks: "))
        students[roll] = {"name": name, "marks": marks}
        save_students(students)
        print("Student updated successfully!")
    else:
        print("Student not found!")


def delete_student():
    students = load_students()
    roll = input("Enter Roll Number to delete: ")

    if roll in students:
        del students[roll]
        save_students(students)
        print("Student deleted successfully!")
    else:
        print("Student not found!")
```

```python
import student_module as sm

while True:
    print("\n----- Student Management System -----")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Student")
    print("5. Delete Student")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        sm.add_student()
    elif choice == "2":
        sm.view_students()
    elif choice == "3":
        sm.search_student()
    elif choice == "4":
        sm.update_student()
    elif choice == "5":
        sm.delete_student()
    elif choice == "6":
        print("Exiting program...")
        break
    else:
        print("Invalid choice, try again!")
```
