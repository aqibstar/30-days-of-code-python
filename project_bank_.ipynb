##1. KPI calculation function


```python
def calculate_kpis(dataframe):
    total_customers = dataframe['Customer_ID'].nunique()
    approved_amount = dataframe[dataframe['Loan_Status'] == 'Approved']['Loan_Amount'].sum()
    total_outstanding = dataframe['Outstanding_Balance'].sum()
    approval_rate = (dataframe['Loan_Status'] == 'Approved').mean() * 100
    return total_customers, approved_amount, total_outstanding, approval_rate

calculate_kpis(df)
```

###2. Group-wise analysis

```python
def analyze_by_group(dataframe, group_col):
    return dataframe.groupby(group_col)[['Loan_Amount', 'Outstanding_Balance']].mean()

analyze_by_group(df, 'Loan_Type')
```

###3. Outlier detection

```python
def find_outliers(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    threshold = Q3 + 1.5 * IQR
    return dataframe[dataframe[column] > threshold]

find_outliers(df, 'Outstanding_Balance')
```

```python
class BankAnalytics:
    def __init__(self, dataframe):
        self.df = dataframe

    def calculate_kpis(self):
        total_customers = self.df['Customer_ID'].nunique()
        approved_amount = self.df[self.df['Loan_Status'] == 'Approved']['Loan_Amount'].sum()
        total_outstanding = self.df['Outstanding_Balance'].sum()
        approval_rate = (self.df['Loan_Status'] == 'Approved').mean() * 100
        return total_customers, approved_amount, total_outstanding, approval_rate

    def loan_analysis(self, group_col='Loan_Type'):
        return self.df.groupby(group_col)[['Loan_Amount', 'Outstanding_Balance']].mean()

    def branch_analysis(self, branch_df, group_col='Branch'):
        return branch_df.groupby(group_col)[['Loan_Amount', 'Outstanding_Balance']].sum()

    def credit_analysis(self, credit_df):
        return credit_df.groupby('Credit_Score_Group')['Loan_Status'].apply(
            lambda x: (x == 'Approved').mean() * 100
        )

    def search_customer(self, customer_id):
        return self.df[self.df['Customer_ID'] == customer_id]
```

```python
bank = BankAnalytics(df)

bank.calculate_kpis()
bank.loan_analysis('Employment_Type')
bank.branch_analysis(df_branch)
bank.credit_analysis(df_credit)
bank.search_customer('CUST00123')
```
