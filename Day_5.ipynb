```python
#1.Hollow Square Pattern
for i in range(4):
  for j in range(4):
   if i==0 or i==3 or j==0 or j==3:
      print("*", end="")
   else:
     print(" ", end="")
  print()
```

```python
#2.Hollow Rectangle
for i in range(5):
  for j in range(5):
    if i==0 or i==4 or j==0 or j==4:
       print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#3.Hollow Right Triangle
n=5
for i in range(n+1):
  for j in range(1,i+1):
    if i==1 or i==n or j==1 or j==i:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
for i in range(5,0
               ,-1):
  for j in range(i+1):
    if i==5 or j==1 or j==i:
       print("*",end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#5.Pyramid Pattern
row = 5
for i in range(row+1):
    for j in range(row - i):
        print(" ", end = "")
    for k in range(2 * i - 1):
        print("*", end = "")
    print()
```

```python
for row in range(1,6):
  for col in range(1,10):
    if row == 5 or row+col==6 or col-row==4:
      print("*", end=" ")
    else:
      print(" ", end=" ")
  print()
```

```python
#Diamond triangle
row = 9
for i in range(row+1):
  for i2 in range(row,0,-1):
    for j in range(row - i):
      for j2 in range(5-i2):
        print(" ", end = "")
    for k in range(2 * i - 1):
      for l in range( 2 * i2 +1):
        print("*", end = "")
    print()
```

```python
row=5
for i in range(1,row+1):
  for j in range(row - i):
      print("", end=" ")
  for k in range (1,2 * i):
      print(k,end="")
  print()
```

```python
row=5
for i in range(1,1+row):
  for j in range(row-i):
    print("", end=" ")
  for k in range(1,2*i):
    print(i, end="")
  print()
```

```python
# Palindrome Number Pyramid
n=5
for i in range(n+1):
  for j in range(n-i):
    print("", end=" ")
  for k in range(1,i+1):
    print(k,end="" )
  for k in range(i-1,0,-1):
    print(k,end="")
  print()
```

```python
n = 5
for i in range(n, 0, -1):
  for j in range(n - i):
    print(" ", end="")
  for k in range(1, i + 1):
    print(k, end="")
  for k in range(i - 1, 0, -1):
    print(k, end="")
  print()
```
